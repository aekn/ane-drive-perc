# YOLO11s Baseline; BDD100K 10K Detection Subset

This notebook trains a YOLO11s detector on the `train_010k_seed42` BDD100K detection subset and evaluates on the full `val_full` split.

Purpose is to establish a stronger pretrained YOLO baseline than nano models while keeping the input size at `640` for standard comparison. May serve as baseline teacher for our ANE model.

```yaml
experiment:
    task: detection
    model: yolo11s.pt
    dataset: bdd100k-det
    train_split: train_010k_seed42
    val_split: val_full
    image_size: 640
    rectangular_batches: true
    epochs: 50
    save_dir: MyDrive/projects/ane-drive-perc/runs/baselines/yolo11s_bdd010k_rect640
```
